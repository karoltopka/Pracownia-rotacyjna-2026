```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import pdist
from sklearn.preprocessing import StandardScaler

df = pd.read_excel('deskryptory_z_cas_2.xlsx', dtype={0: str})

cas = df.iloc[:, 0].values
dane = df.iloc[:, 1:].values

scaler = StandardScaler()
dane_znormalizowane = scaler.fit_transform(dane)

odleglosci = pdist(dane_znormalizowane, metric='euclidean')
z = linkage(odleglosci, method='ward')
    
plt.figure(figsize=(14, 7))
dendrogram(z, labels=cas, leaf_rotation=90)
plt.title('HCA')
plt.xlabel('CAS')
plt.ylabel('odległość euklidesowa')
plt.show()
```
